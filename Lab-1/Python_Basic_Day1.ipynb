# Variables in Python

Python variable names must follow certain rules to make the code readable and error-free.

## Rules for Naming Variables

### 1. Variable Name Must Start with a Letter or Underscore

A variable name can start with:
- Uppercase letter (`A-Z`)
- Lowercase letter (`a-z`)
- Underscore (`_`)

Example:

```python
name = "Rahim"
_age = 25
Student = "Ali"
```

---

### 2. Variable Name Cannot Start with a Number

A variable name cannot begin with digits.

Incorrect:

```python
1name = "Rahim"
25age = 20
```

Correct:

```python
name1 = "Rahim"
age25 = 20
```

---

### 3. Variable Names Can Contain Only Letters, Numbers, and Underscores

Allowed characters:

- Letters (`A-Z`, `a-z`)
- Numbers (`0-9`)
- Underscore (`_`)

Example:

```python
student_name = "Karim"
student1 = "Rahim"
```

---

### 4. Spaces Are Not Allowed in Variable Names

Incorrect:

```python
student name = "Ali"
```

Correct:

```python
student_name = "Ali"
```

Use underscore (`_`) instead of spaces.

---

### 5. Special Characters Are Not Allowed

Characters like:

```
@, #, $, %, -, &
```

cannot be used in variable names.

Incorrect:

```python
student-name = "Ali"
price$ = 500
```

Correct:

```python
student_name = "Ali"
price = 500
```

---

### 6. Variable Names Are Case-Sensitive

Python treats uppercase and lowercase letters as different.

Example:

```python
Age = 20
age = 30
```

Here, `Age` and `age` are two different variables.

---

### 7. Keywords Cannot Be Used as Variable Names

Python reserved words cannot be used as variable names.

Examples of keywords:

```
if
else
for
while
class
def
return
True
False
```

Incorrect:

```python
class = "Python"
if = 10
```

---

## Best Practices for Variable Naming

- Use meaningful variable names.
- Use lowercase letters for normal variables.
- Use underscores to separate multiple words.
- Avoid using single letters except for simple counters.

Good Examples:

```python
student_name = "Rahim"
total_marks = 450
average_score = 90.5
```

Poor Examples:

```python
x = "Rahim"
a = 450
```

---

## Summary Table

| Rule | Example |
|---|---|
| Start with letter or underscore | `name`, `_age` |
| Cannot start with number | ❌ `1name` |
| Only letters, numbers, underscore allowed | `student1` |
| No spaces allowed | ❌ `student name` |
| No special characters | ❌ `price$` |
| Case-sensitive | `Age` ≠ `age` |
| Cannot use keywords | ❌ `class` |

# Multiple Variable Assignment in Python

Python allows assigning values to multiple variables in a single line. This feature is called **multiple variable assignment**.

It helps to write cleaner and shorter code.

## Syntax

```python

# Types of Python Operators

Operators are special symbols or keywords used to perform different operations on variables and values.

Python operators are divided into **7 main categories**:

1. Arithmetic Operators
2. Assignment Operators
3. Comparison Operators
4. Logical Operators
5. Identity Operators
6. Membership Operators
7. Bitwise Operators

---

# 1. Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations.

These operators work with numeric values such as integers and floating-point numbers.

| Operator | Name | Description | Example |
|---|---|---|---|
| `+` | Addition | Adds two values | `10 + 5` |
| `-` | Subtraction | Subtracts one value from another | `10 - 5` |
| `*` | Multiplication | Multiplies two values | `10 * 5` |
| `/` | Division | Divides one value by another | `10 / 5` |
| `%` | Modulus | Returns the remainder | `10 % 3` |
| `**` | Exponent | Returns power value | `2 ** 3` |
| `//` | Floor Division | Returns integer division result | `10 // 3` |

---

# 2. Assignment Operators

Assignment operators are used to assign values to variables.

They are also used to update the existing value of a variable.

| Operator | Description | Example |
|---|---|---|
| `=` | Assign value | `x = 10` |
| `+=` | Add and assign | `x += 5` |
| `-=` | Subtract and assign | `x -= 5` |
| `*=` | Multiply and assign | `x *= 5` |
| `/=` | Divide and assign | `x /= 5` |
| `%=` | Modulus and assign | `x %= 5` |

---

# 3. Comparison Operators

Comparison operators are used to compare two values.

The result of comparison operators is always a Boolean value:

- `True`
- `False`

| Operator | Name | Example |
|---|---|---|
| `==` | Equal to | `a == b` |
| `!=` | Not equal to | `a != b` |
| `>` | Greater than | `a > b` |
| `<` | Less than | `a < b` |
| `>=` | Greater than or equal | `a >= b` |
| `<=` | Less than or equal | `a <= b` |

---

# 4. Logical Operators

Logical operators are used to combine multiple conditions.

They are commonly used with decision-making statements.

| Operator | Description | Example |
|---|---|---|
| `and` | Returns True if all conditions are True | `a > 5 and b < 10` |
| `or` | Returns True if at least one condition is True | `a > 5 or b < 10` |
| `not` | Reverses the result | `not(a > 5)` |

---

# 5. Identity Operators

Identity operators are used to check whether two variables refer to the same object in memory.

| Operator | Description | Example |
|---|---|---|
| `is` | Returns True if both variables are the same object | `x is y` |
| `is not` | Returns True if variables are different objects | `x is not y` |

---

# 6. Membership Operators

Membership operators are used to check whether a value exists inside a collection.

They are commonly used with:

- String
- List
- Tuple
- Set
- Dictionary

| Operator | Description | Example |
|---|---|---|
| `in` | Checks if a value exists | `"Python" in skills` |
| `not in` | Checks if a value does not exist | `"Java" not in skills` |

---

# 7. Bitwise Operators

Bitwise operators perform operations on binary numbers.

They work at the bit level.

| Operator | Name | Description |
|---|---|---|
| `&` | AND | Sets bit to 1 if both bits are 1 |
| `|` | OR | Sets bit to 1 if any bit is 1 |
| `^` | XOR | Sets bit to 1 if bits are different |
| `~` | NOT | Reverses bits |
| `<<` | Left Shift | Shifts bits to the left |
| `>>` | Right Shift | Shifts bits to the right |

---

# Summary of Python Operators

| Operator Category | Main Purpose |
|---|---|
| Arithmetic | Perform mathematical calculations |
| Assignment | Store and update values |
| Comparison | Compare two values |
| Logical | Combine conditions |
| Identity | Compare object identity |
| Membership | Check existence in collections |
| Bitwise | Perform binary operations |

````python
# Arithmetic Operators Example

a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus (Remainder):", a % b)
print("Exponent:", a ** b)
print("Floor Division:", a // b)
````

````python
# Example

monthly_salary = 40000
bonus = 5000

total_income = monthly_salary + bonus

print("Monthly Salary:", monthly_salary)
print("Bonus:", bonus)
print("Total Income:", total_income)
````

````python
# Assignment Operators Example

production = 1000

print("Initial Production:", production)

production += 500
print("After Adding:", production)

production -= 200
print("After Deduction:", production)

production *= 2
print("After Multiplication:", production)

production /= 2
print("After Division:", production)
````

````python
# Comparison Operators Example

age = 27

print("Age greater than 18:", age > 18)
print("Age equal to 27:", age == 27)
print("Age not equal to 30:", age != 30)
print("Age less than 50:", age < 50)
print("Age greater or equal to 25:", age >= 25)
print("Age less or equal to 20:", age <= 20)
````

````python
# Logical Operators Example

age = 27
has_id_card = True

print("Can Enter:", age >= 18 and has_id_card)

print("Age Valid or Has ID:", age >= 18 or has_id_card)

print("ID Not Available:", not has_id_card)
````

````python
# Identity Operators Example

list1 = [1, 2, 3]
list2 = list1
list3 = [1, 2, 3]

print("list1 is list2:", list1 is list2)
print("list1 is list3:", list1 is list3)

print("list1 is not list3:", list1 is not list3)
````

````python
# Identity Operators

x = ["Python", "C++"]
y = x
z = ["Python", "C++"]

print(x is y)
print(x is z)

print(x is not z)
````

````python
# Membership Operators

courses = ["Python", "C++", "Java"]

print("Python" in courses)
print("HTML" in courses)

print("HTML" not in courses)
````

````python
# Membership Operator with String

skills = ["Python", "Machine Learning", "Data Science"]

print("Python" in skills)
print("SQL" in skills)
print("SQL" not in skills)
````
