### Create a list containing the marks of 5 courses. Display the total marks, average mark, highest mark, and lowest mark. Also count how many courses the student passed (pass mark = 50).

```python
programming_ds = float(input("Enter your Programming for DS mark: "))
data_visualization = float(input("Enter your Data Visualization mark: "))
data_processing = float(input("Enter your Data Processing mark: "))
data_science = float(input("Enter your Data Science mark: "))
java = float(input("Enter your Java mark: "))

marks = [programming_ds, data_visualization, data_processing, data_science, java]

i = 0
total = 0
highest = marks[0]
lowest = marks[0]
passed = 0
failed = 0

for mark in marks:
    total = total + marks[i]

    if marks[i] > highest:
        highest = marks[i]

    if marks[i] < lowest:
        lowest = marks[i]

    if marks[i] >= 50:
        passed += 1
    else:
        failed += 1

    i += 1

average = total / len(marks)

print()
print("Total Marks:", total)
print("Average Mark:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Passed Courses:", passed)
print("Failed Courses:", failed)
```

### Create a tuple containing the temperatures recorded on Saturday and Sunday for three different times of the day. Use a for loop to display each temperature. Then calculate the highest and lowest temperature in the tuple.

```python
temperatures = (28, 32, 30, 27, 31, 29)

print("Recorded Temperatures:")
for temp in temperatures:
    print(temp, "°C")

highest = max(temperatures)
lowest = min(temperatures)

print("Highest Temperature:", highest, "°C")
print("Lowest Temperature:", lowest, "°C")
```

### Ask the user to enter a word or short sentence. Display the number of characters, the number of vowels (a, e, i, o, u), and the number of spaces. Use a for loop to examine each character.

```python
text = input("Enter a word or short sentence: ")

characters = len(text)
vowels = 0
spaces = 0

for ch in text:
    if ch.lower() in "aeiou":
        vowels += 1
    if ch == " ":
        spaces += 1

print("Number of characters:", characters)
print("Number of vowels:", vowels)
print("Number of spaces:", spaces)
```

### Write a function called calculate_grade(marks) that receives a list of marks. The function should calculate the average and return a grade using these rules: A for average ≥ 80, B for ≥ 70, C for ≥ 60, D for ≥ 50, and F for below 50. Call the function using a list of 5 marks.

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


marks = [85, 48, 91, 67, 73]

print("Grade:", calculate_grade(marks))
```

### Create a list of dictionaries representing 4 products. Each dictionary should contain product name, price, and quantity. Use a for loop to calculate the total cost of the cart. Then write a function called apply_discount(total) that returns the final price after a 10% discount when the total is greater than or equal to 2,000; otherwise, no discount is applied.

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total = 0

for product in products:
    total = total + product["price"] * product["quantity"]

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

print("Total cost:", total)
print("Final price:", apply_discount(total))
```

### Write a program using a while loop that asks the user for a number and prints a countdown from that number to zero.

```python
num = int(input("Enter a number: "))

while num >= 0:
    print(num)
    num = num - 1
```
